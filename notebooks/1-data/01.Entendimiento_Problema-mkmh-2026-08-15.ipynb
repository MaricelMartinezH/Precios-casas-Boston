## 🔎Entendimiento del problema

### ¿Cuál es el objetivo del problema?

El objetivo del proyecto es desarrollar un modelo de aprendizaje automático que permita predecir el valor medio de las viviendas (medv) a partir de diferentes características socioeconómicas, ambientales y relacionadas con las viviendas y su entorno en diferentes zonas de Boston.

La variable medv corresponde a la variable objetivo que se desea estimar, mientras que las demás variables, excepto ID, se consideran inicialmente como variables predictoras candidatas.

### ¿Cómo se usará su solución?

La solución se utilizará para realizar estimaciones del valor medio de las viviendas a partir de las características disponibles de cada registro.

El modelo podrá servir como una herramienta de apoyo para estimar precios y analizar la relación entre las características socioeconómicas, ambientales y estructurales de una zona y el valor de las viviendas.

La predicción obtenida por el modelo debe entenderse como una estimación basada en los datos disponibles y no como una valoración inmobiliaria definitiva.

### ¿Cuáles son las soluciones actuales?

Actualmente, la estimación del valor de una vivienda puede realizarse mediante métodos tradicionales de valoración inmobiliaria y mediante modelos estadísticos y de aprendizaje automático.

Para este proyecto se plantea utilizar aprendizaje automático con el fin de aprender automáticamente la relación entre las características de las zonas y el valor medio de las viviendas.

### ¿Cómo se debe enmarcar este problema?

El problema se plantea como un problema de aprendizaje supervisado, debido a que se dispone de una variable objetivo conocida (medv) que se desea predecir.

Es específicamente un problema de regresión, ya que la variable objetivo es numérica.

Inicialmente se considera un problema fuera de línea (offline), debido a que se trabajará con un conjunto de datos histórico y estático. El modelo será entrenado utilizando los datos disponibles y posteriormente será evaluado sobre datos que no hayan sido utilizados durante el entrenamiento.

### ¿Cómo medir el desempeño?

Al tratarse de un problema de regresión, se utilizarán métricas adecuadas para comparar los valores reales de medv con los valores predichos por el modelo.

Como primera aproximación se consideran:

MAE (Mean Absolute Error): mide el error absoluto promedio de las predicciones.
RMSE (Root Mean Squared Error): mide la magnitud de los errores y da mayor peso a errores grandes.
R²: indica qué proporción de la variabilidad de la variable objetivo puede ser explicada por el modelo.

Como métrica principal se propone inicialmente utilizar MAE, complementada con RMSE y R².

### ¿La medida está alineada con el objetivo?

Sí. La métrica propuesta está alineada con el objetivo porque el propósito del modelo es obtener estimaciones de medv lo más cercanas posible a sus valores reales.

El MAE permite medir directamente la magnitud promedio del error de las predicciones y se expresa en las mismas unidades de la variable objetivo. El RMSE complementa esta evaluación al dar mayor importancia a los errores grandes, mientras que R² permite conocer la capacidad explicativa del modelo.

### ¿Cuál es el rendimiento mínimo necesario?

En esta etapa no es posible establecer un valor numérico definitivo para el desempeño mínimo requerido, debido a que todavía no se ha construido ni evaluado un modelo.

El desempeño mínimo aceptable se determinará después de comparar los diferentes modelos y analizar el error obtenido en relación con el objetivo de predicción.
